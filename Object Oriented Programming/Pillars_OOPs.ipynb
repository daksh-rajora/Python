# Pillars of Object Oriented Programming

## Inheritance
#### Inheritance is a Fundamental concept in Object Oriented Programming because of which the child class is able to inherit the properties OR Attributes and functionalites of the parent class

#### Just like a child is going to inherit the properties from parent and parent is going to inherit the properties from the grandparent and definitely in this hierarchy the grandchild can have some features different from the parent and grandparent as well.

#### In Python typically we got to see 3 Types of Inheritance:

#### 1. Single Inheritance
#### 2. Multilevel Inheritance
#### 3. Multiple Inheritance

## Single Inheritance ---> The features or the attributes from one parent class will be getting inherited into one child class

#### Parent Class (Properties ---> Attributes and Functionalities)
#### |
#### |
#### Child Class

## Multilevel Inheritance ---> The properties or methods and attributes of the grandparent class can be inherited by the parent class and then the properties of the parent class will be inherited by the child class. In a proper hierarchy the inheritance process is going to work for us.

### Grandparent (Properties)
### |
### |
### Parent(Properties)
### |
### |
### child

## Multiple Inheritance --> In Multiple Inheritance, The properties of two parent class or multiplw parent class will be getting inherited into the child class.

### Parent_1(Properties) + Parent_2(Properties) + Parent_3(Properties).....
### |
### |
### Child

```python
class Avenger:

    def __init__(self,name,power):
        self.name=name
        self.power=power

    def introduction(self):
        print(f"I am {self.name} and my power is {self.power}")

class Ironman(Avenger):

    def __init__(self, name, power, weapon):
        super().__init__(name, power)
        self.weapon=weapon

    def weapon_details(self):
        print(f"My weapon is {self.weapon}")
```

```python
iron=Ironman("Iron Man", "Genius Intellect", "Repulsor Beam")
```

```python
class Grandparent:

    def __init__(self,nationality,baldness,anger_issues):
        self.nationality=nationality
        self.baldness=baldness
        self.anger_issues=anger_issues

    def introduction(self):
        print(f"I am {self.nationality} and I do have baldness issue and anger issues is {self.anger_issues}")
```

```python
class Parent(Grandparent):

    def __init__(self, nationality, baldness, anger_issues, intelligence):
        super().__init__(nationality, baldness, anger_issues)
        self.intelligence=intelligence

    def intelligence_details(self):
        print(f"About my intelligence if you guys want to know that whether if i am intelligent or not then my answer is: {self.intelligence}")
```

```python
class Child(Parent):

    def __init__(self, nationality, baldness, anger_issues, intelligence, charm):
        super().__init__(nationality, baldness, anger_issues, intelligence)
        self.charm=charm

    def charm_details(self):
        print(f"About my charm if you guys want to know that whether if i am charming or not then my answer is: {self.charm} ")
```

```python
parent_1=Parent("Indian","Yes","No","Yes")
```

```python
parent_1.introduction()
```

```python
child_1=Child("Indian","No","Yes","Yes","Yes")
```

```python
child_1.introduction()
```

```python
child_1.intelligence_details()
```

```python
child_1.nationality="Christan"
```

```python
child_1.introduction()
```

```python
class Ironmans:
     
    def __init__(self,name,power,suit,weapon,nationality="American"):
        self.name=name
        self.power=power
        self.weapon=weapon
        self.nationality=nationality

    def introduction(self):
        print(f"I am {self.name} and my power is {self.power}")

class Captainamerica:

    def __init__(self, name, power, suit, weapon, nationality="American"):
        self.name=name
        self.power=power
        self.weapon=weapon
        self.nationality=nationality

    def weapon_details(self):
        print(f"I carry the weapon : {self.weapon}")
```

```python
class Hulk(Ironmans, Captainamerica):

    def __init__(self, name, power, suit, weapon, nationality, anger_issues):
        super().__init__(name, power, suit, weapon, nationality)
        self.anger_issues=anger_issues

    def anger_details(self):
        print(f"I am having anger issues so my answer is: {self.anger_issues}")
```

```python
hulk=Hulk("Hulk", "Amazing Power and Strength", "Mark 42 Ironman Suit","Repulsor Beams and Shield","American","Yes")
```

```python
hulk.introduction()
```

```python
hulk.weapon_details()
```

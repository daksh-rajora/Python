# Encapsulation

## Encapsultaion is a way of wrapping data (Variables and Methods) inside a capsule or a class as a single unit.

## It restricts the direct access to some of the objects component's which is a means of preventing accidental interferance and misuse of the data.

# Key concepts of Encapsulation:
## 1. Public Members
## 2. Protected Members
## 3. Private Members
## 4. Getter and Setter methods

## Public Member --> Accessible from anywhere or attributes are accessible outside of the class and these members are accessible by the derived class.
## Public member attribute is directly declared

```python
class Ironman:

    def __init__(self,name,power,weapon):
        self.name=name
        self.power=power
        self.weapon=weapon

    def introduction(self):
        print(f"Hello my name is {self.name} and i have weapon {self.weapon}")
```

```python
## dir()---> it takes the input as the object name and gives us this idea that which member is public or private.

iron=Ironman("Tony Stark", "Genius Level Intelligence", "Armor")

dir(iron)
```

```python
iron.name

```

```python
iron.introduction()
```

## Private Members---> A Privatr member is not accessible outside the class and not even in the derived class in order to have the access over these private members or to bring some modification on these members we use getter and setter method.
## Private members are declared by using double underscore __ before the name

```python
class Avenger:

    def __init__(self,name,phone_number):
        self.name=name
        self.__phone_number=phone_number
```

```python
aveng=Avenger("Natasha", 9876543210)
```

```python
aveng.name
```

```python
aveng.phone_number
```

```python
dir(aveng)
```

```python
## getter method -->it will be helping us out with accessing the private data

class Captain:

    def __init__(self,name,address):
        self.name=name
        self.__address=address

    ## getter method
    def accessing_add(self):
        print(f"My address is {self.__address}")
```

```python
captain=Captain("Stieve Roggers","New York")
```

```python
captain.name
```

```python
captain.address
```

```python
captain.accessing_add()
```

```python
## setter method --> it will help us in modifying the private data

class Animal:

    def __init__(self,name,sound):
        self.name=name
        self.__sound=sound

    def sound_info(self): ##getter method
        print(f"The Animal is going to sound as: {self.__sound}")

    def sound_modify(self, new_sound): ##setter method
        self.__sound=new_sound
        print(f"The modified sound is : {new_sound}")
```

```python
animal=Animal("Dog", "Bow Bow")
```

```python
animal.sound
```

```python
animal.sound_info()
```

```python
animal.sound_modify("meow meow")
```

## Protected Data is the data which is not accessible outside the class but the protected data is going to be accessible by the derived classor child class
## For having the protected member you can mention your protected member as _attribute or by using a single underscore _

```python
class Grandparent:

    def __init__(self,name,locker_key):
        self.name=name
        self._locker_key=locker_key

    ## getter method
    def locker_info(self):
        print(f"The locker key is {self._locker_key}")
```

```python
class Parent(Grandparent):

    def __init__(self,name,locker_key):
        super().__init__(name,locker_key)

    def child_reveals(self):
        print(f"The locker key is {self._locker_key}")
```

```python
grandpa=Grandparent("John", "1234")
```

```python
grandpa.name
```

```python
grandpa.locker_key
```

```python
parent=Parent("Max","3243")
```

```python
parent.locker_key
```

```python
parent.child_reveals()
```

```python
# Bank Account System
class BankAccount:

    def __init__(self, account_holder, balance):
        self.account_holder=account_holder
        self.__balance=balance

    def deposit(self, amount):
        if amount>0:
            self.__balance+=amount
            print(f"Deposited ${amount}. New balance is ${self.__balance}")
        else:
            print("Invalid deposit amount")

    def withdraw(self, amount):
        if 0<amount<=self.__balance:
            self.__balance-=amount
            print(f"Withdrawn ${amount}. New balance is ${self.__balance}")
        else:
            print("Insufficient funds or invalid withdrawal amount")

    def get_balance(self):
        return self.__balance
```

```python
account=BankAccount("Alice", 100000)
```

```python
account.account_holder
```

```python
account.balance
```

```python
account.get_balance()
```

```python
account.withdraw(89843)
```

```python
class Avenger:

    def __init__(self, name, power, weapon, nationality="American"):
        self.name=name
        self.power=power
        self.weapon=weapon
        self.nationality=nationality

    def introduction(self):
        print(f"Hello my name is {self.name}")

    def carrying_weapon(self):
        print(f"I am carrying {self.weapon}")
```

## Types of Constructors:
## 1. Default Constructor
## 2. Parameterized Constructor

## 1. Default Constructor: It does not have the parameters inside it and it is only having the self keyword inside it. It assigns a default value to all the objects

## 2. Parameterized Constructor: It consists of parameters inside it so that every object can have specific attribute associated to it
## It allows us to pass arguments while creating an object.

```python
class Dog:

    def __init__(self):
        self.breed="Labrador"
        self.color="White"
        self.eye_color="Black"

    def show_info(self):
        print(f"The dog is {self.breed} and its color is {self.color}")
```

```python
dog1=Dog()
```

```python
dog1.breed
```

```python
dog1.show_info()
```

```python
class Dogs:

    def __init__(self,breed,color):
        self.breed=breed
        self.color=color

    def show_info(self):
        print(f"The dog is {self.breed} and its color is {self.color}")
```

## Types of Methods
## 1. Instance Method: Instance Method operate on Object Attributes.
## The instance method is going to use self keyword to access and modify object attributes
## These Methods are called using an object.

## 2. Class Method ----> It is the method of the whole class or the whole blueprint which will remain the same for all of the objects then such kind of methods is being reffered as a class method.
## In class method, these methods work with the classs level attributes.
## They are going to use cls istead of self
## They are callled using the class name as well
## For specific that the currend method is a class method we have to use the abbreviation as @classmethod

## 3. Static Method---> static method do not use self or cls.
## they work like normal functions but belong to a class.
## They are going to be designed by a abbreviation @staticmethod
## static method can also be called by the classname

```python
class Avengers:

    def __init__(self, name, power, weapon, nationality="American"):
        self.name=name
        self.power=power
        self.weapon=weapon
        self.nationality=nationality

    def show_details(self):
        print(f"My name is {self.name} and i am having power {self.power}")
```

```python
iron_man=Avengers("Iron Man", "Genius Intellect and Technology", "Repulsor beams")
```

```python
iron_man.show_details()
```

```python
Thor=Avengers("Thor", "God of Thunder", "Mjolnir", "Asgardian")
```

```python
Thor.show_details()
```

```python
class Avengerss:

    origin="Avenger"

    def __init__(self,name,power):
        self.name=name
        self.power=power

    @classmethod
    def show_origin(cls):
        print(f"The origin of avengers is {cls.origin}")

    def self_info(self):
        print(f"My name is {self.name} and i have {self.power}")
```

```python
Iron_man=Avengerss("Iron Man", "Genius Intellect and Technology")
```

```python
Iron_man.self_info()
```

```python
Iron_man.show_origin()
```

```python
class Avengersss:

    @staticmethod
    def mission():
        print("All Avengers must protect the Earth")

    origin="Avenger"

    def __init__(self, name, power):
        self.name=name
        self.power=power

    @classmethod
    def show_origin(cls):
        print(f"The origin of avengers is {cls.origin}")

    def introduction(self):
        print(f"My name is {self.name} and i have {self.power}")
```

```python
iron_man=Avengersss("Iron Man", "Genius Intellect and Technology")
```

```python
iron_man.introduction()
```

```python
iron_man.mission()
```

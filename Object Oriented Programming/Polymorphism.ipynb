# Polymorphism

## Polymorphism means MANY FORMS in Greek. 
## In OOPs, Ploymorphism allows objects of different classes as if they are the object of the same class & It enables reusability and makes programs mode flexible.

## Types of Polymorphism:
### 1. Method Overriding (Runtime Polymorphism)
### 2. Method Overloading (Compile Time Polymorphism)
### 3. Operator Overloading

## Method Overriding--> A child class provides a specific implementation of a method that has already been defined into the parent class. This allows Dynamic selection at Runtime.

```python
class Grandparent: #Parent class

    def speak(self):
        print("Hello My name is Max")

class Parent(Grandparent): #Child class

    def speak(self):
        print("Hello My name is John")
        
class Child(Grandparent): #Child class

    def speak(self):
        print("Hello My name is Alice")
```

```python
def attribute_provider(objects):
    return objects.speak() #Method Overriding
    
```

```python
grandpa=Grandparent()
```

```python
attribute_provider(grandpa)
```

```python
parent=Parent()
```

```python
attribute_provider(parent)
```

```python
child=Child()
```

```python

attribute_provider(child)
```

```python
class Shape:
    def area(self):
        return "This method returns the area of the shape"
```

```python
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width=width
        self.height=height

    def area(self):
        return self.width*self.height
```

```python
class Circle(Shape):

    def __init__(self, radius):
        self.radius=radius

    def area(self):
        return 3.14*self.radius**2
```

```python
shape=Shape()
```

```python
shape.area()
```

```python
rectangle=Rectangle(6,5)
rectangle.area()
```

```python
circle=Circle(4)
circle.area()
```

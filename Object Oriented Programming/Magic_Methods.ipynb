# Magic Methods

## Magic Methods are also called as Dunder Methods because they have double underscores__ before and after their name and they are special methods in python that allow us to customize how objects behave.

## These methods can help us define:
## 1. Object Initialization (__init__)
## 2. String Representation (__str__, __repr__)
## 3. Artihmetic Operations (__add__, __sub__,etc)
## 4. Comparison Operations (__eq__,__lt__)
## 5. Length, indexing, iteration (__len__, __getitem__,etc)

## Why to use Magic Methods

## 1. Makes objects behave like built in types (eg,lists,integers)
## 2. Improves Readability by allowing basics operations object1 + object2
## 3. Simplifies code by replacing verbose function calls with natural syntax.  

## print(object) --> __str__ --> what is returned when print(object) is used
## Arithmetic operations can be done among the objects by using Magic Methods

```python
## __add__

class Avenger:

    def __init__(self,name,strength):
        self.name = name
        self.strength = strength

    def __add__(self, other):
        return Avenger(self.name + "&" + other.name, self.strength + other.strength)

    def __str__(self):
        return f"{self.name} with combined strength of  {self.strength}"
ironman=Avenger("Ironman", 250)
thor=Avenger("Thor",100)

team=ironman + thor
print(team)
```

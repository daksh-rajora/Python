# Abstraction

## Abstraction is the concept of hiding the implementation details and only showing the necessary features of the object.
## This helps in reducing programming complexity and efforts
## Abstration allows us to work with an object with what an object does rather than how it does it.


### How can we achieve abstraction in Python
## 1. Abstract class --> A class that cannot be instantiated or that means for an abstract class we cant create the object of this class.
## 2. Abstract method--> Abstract method just beacuse it cant accessed by the object of the abstract class itself thats why we have to call this abstract method in our child class and the implementation of the abstract method is not provided inside the abstract class but must be provided in the child class.
## The abstract method is being defined by @abstractmethod

```python
from abc import ABC,abstractmethod

class Vechile(ABC):

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

class Car(Vechile):

    def start(self):
        print("Car is starting")

    def stop(self):
        print("Car is stopping")

class ElectricScooter(Vechile):

    def start(self):
        print("Electric scooter is starting")

    def stop(self):
        print("Electric scooter is stopping")
```

```python
car=Car()
```

```python
car.start()
```

```python
scooter=ElectricScooter()
```

```python
scooter.start()
```
